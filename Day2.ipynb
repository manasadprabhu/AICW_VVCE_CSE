```python
#__init__ constructor - used to initialize
class person:
    def __init__(self,name,age,school):     #self is used to instantiate other varibles
        self.name = name
        self.age = age
        self.school = school
    def myfun(self):
        print('Hello,My name is',self.name)
        print('My age is',self.age)
        print('I went to',self.school)
```

```python
p1= person("Manasa",20,"BGS")
p1.myfun()
```

```python
# accessing attributes and methods
class emp:
    name = "RMS"
    salary = '25000'
    def show(self):
        print(self.salary)
        print(self.name)
```

```python
e1 = emp()
```

```python
#using get attribute instead e1.name
print(getattr(e1,'name'))
```

```python
# hasattr - return true if the object has attribute
print(hasattr(e1,'name'))        #hasattr is used to check whether a particular attributes is present
```

```python
print(hasattr(e1,'gender'))
```

```python
#setattr - sets an attribute
setattr(e1 , 'gender','F')
```

```python
print(getattr(e1,'gender'))
```

```python
print(hasattr(e1,'gender'))
```

```python
# delattr - del the attribute
delattr(e1,'gender')
```

```python
class Student:
    def __init__(self,usn,name,age):   #constructor
        #Instance variable
        self.usn = usn
        self.name = name
        self.age = age

    #instance method access instance variable
    def show(self):
        print('USN:',self.usn,',Name:',self.name,',Age:',self.age)

    #instance method to modify instance variable 
    def update(self,usn,age):
        self.usn=usn
        self.age=age
        
        
```

```python
#create an object
stu=Student(157,'Manasa',20)
stu.show()
```

```python
stu.update(12,19)
stu.show()
```
